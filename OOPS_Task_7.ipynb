Smart Light Controller

```python
class SmartLight:
    def __init__(self, light_name):
        self.light_name = light_name
        self.status = "OFF"   # Default status

    def turn_on(self):
        self.status = "ON"

    def turn_off(self):
        self.status = "OFF"

    def display_status(self):
        print(f"{self.light_name} is {self.status}")


# 📥 User Input
light_name = input("Enter Light Name: ")
action = input("Enter Action (ON/OFF): ").upper()

# Create object
light = SmartLight(light_name)

# Perform action
if action == "ON":
    light.turn_on()
elif action == "OFF":
    light.turn_off()
else:
    print("Invalid Action")

# 📤 Display Output
light.display_status()
```

Employee ID Card System

```python
class Emp_details:
  def __init__(self,Employee_name,Emp_ID,Department):
    self.Employee_name = Employee_name
    self.Emp_ID = Emp_ID
    self.Department = Department

  def display_ID_card(self):
    print("Employee ID card")
    print("Name",self.Employee_name)
    print("ID",self.Emp_ID)
    print("Department",self.Department)

#user inputs
Name=input("Employee_name: ")
ID=input("Emp_ID: ")
Department=input("Department: ")

#creating an object
emp=Emp_details(Name,ID,Department)

#output
emp.display_ID_card()
```

Mobile Contact Record

```python
class Saving_contact:
  def __init__(self,contact_name,phone):
    self.contact_name =contact_name
    self.phone = phone

  def display_contacts(self):
    print("Contact Saved")
    print("Contact name: ",self.contact_name)
    print("Phone: ",self.phone)
#user inputs
Contact_name=input("Enter contact name: ")
phone=input("Enter phone number: ")

#creating object
contact=Saving_contact(Contact_name,phone)

contact.display_contacts()
```

Product Price Tag Generator


```python
class products_price_tag:
  def __init__(self,product_name,price):
    self.product_name=product_name
    self.price=price

  def display_price_tag (self):
    print("Product name: ",self.product_name)
    print("Price:₹",self.price)

#user inputs
product_name=input("Enter product name: ")
price=int(input("Enter price:"))

#object creation
price_tag=products_price_tag(product_name,price)

price_tag.display_price_tag()
```


Movie Rating Display System


```python
class movie_rating:
  def __init__(self,movie_name,rating):
    self.movie_name = movie_name
    self.rating = rating

  def display_movie_details (self):
    print("Movie: ",self.movie_name)
    print("Rating:",self.rating)

#user input
movie_name=input("Enter movie name:")
rating=float(input("Enter rating:"))

#object creation
rate=movie_rating(movie_name,rating)

rate.display_movie_details()
```

Delivery Address Manager


```python
class customer_details:
  def __init__(self,customer_name,address):
    self.customer_name=customer_name
    self.address=address

  def delivery_details(self):
    print("Delivery details")
    print("Customer:",customer_name)
    print("Address:",address)

#user inputs
customer_name=input("Enter customer name: ")
address=input("Enter address:")

#creating object
customer=customer_details(customer_name,address)

customer.delivery_details()
```
